```python
# Inventory alerts with dictionaries and loops

inventory=[
    {"name":"Matcha", "stock": 100, "reorder_level":15},
    {"name":"Mocha Latte", "stock":180, "reorder_level":25},
    {"name": "Croissant", "stock" :400, "reorder_level":50},
    {"name": "Lemon Tart", "stock":18, "reorder_level":20},
    {"name":"Baguette", "stock":5, "reorder_level":10},
]


def restock_report(items):
  """ Print stock status and return a list of products to reoder."""
  to_reorder=[]
  for item in items:
    if item["stock"]<=item["reorder_level"]:
      status="REORDER"
      to_reorder.append(item["name"])
    else:
      status="OK"
    print(f"{item['name']:<6} stock={item['stock']:>3} {status}")
  return to_reorder


print(restock_report(inventory))
```

```python
# Weekly Sales Report for The Chill Cafe & Bakery

#Problem - Records sales in a notebook
# Questions
# 1. How much did I make this week?
# 2. What does a typical customer spend?
# 3. Which days should I focus on improving?

# Step 1: Structure the data.
# Each day is a record, so we use a list of dictionaries

week=[
    {"day": "Mon", "sales":89000, "customers": 75},
    {"day": "Tue", "sales":65000, "customers":56},
    {"day": "Wed", "sales":98000, "customers":125},
    {"day": "Thu", "sales":62000, "customers":46},
    {"day": "Fri", "sales":75000, "customers":67},
    {"day": "Sat", "sales":120000, "customers":213},
    {"day": "Sun", "sales":103000, "customers":158},

]
```

```python
# Building a reusable function
#Instead of scattered calculations, we write one function that can produce this report for any week she gives us in future.

def weekly_reports(records):
  """ Summarise a week of The Chill Cafe & Bakery sales records."""
  total_sales=0
  total_customers=0
  for r in records:
    total_sales +=r["sales"]
    total_customers +=r["customers"]

  average_day=total_sales/len(records)
  avg_basket=total_sales/total_customers

  best_day=records[0]
  weak_days=[]
  for r in records:
    if r["sales"]>best_day["sales"]:
      best_day=r
    if r["sales"]<average_day:
      weak_days.append(r["day"])


  print(f"Total sales: KES {total_sales:,.2f}")
  print(f"Total customers: {total_customers}")
  print(f"Average sales per day: KES {average_day:,.2f}")
  print(f"Avg basket size: KES {avg_basket:,.2f}")
  print(f"Best day: {best_day['day']} with KES {best_day['sales']:,.2f}")
  print(f"Below-average days: {', '.join(weak_days)}")


weekly_reports(week)

```

# The Chill Cafe & Bakery
Made KES 612,000 this week. Each customer spends about KES 828 on average. Weekends are your strongest days, and Tuesday, Thursday, and Friday are all below average.

# Recommendation
The business is doing okay with massive sales per day, consider having specials and discounts on the below-average days to attract more customers.
